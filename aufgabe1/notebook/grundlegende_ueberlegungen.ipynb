
**Aufgabe (a)**

Da die Gleichungen (8), (9), (10) dieselbe Gleichung sind, lösen wir die Gleichung (8) für konstante Spannung:
$$\frac{dn}{dt} = \alpha(1-n) - \beta n = \alpha - n(\alpha + \beta)$$
$$
\frac{dn}{dt} = (\alpha + \beta)\left(\frac{\alpha}{\alpha + \beta}-n\right) = \frac{n_{\infty} - n}{\tau}
\tag{16}
$$
Mit Trennung der Variablen lässt sich $n(t)$ finden:
$$\int_{n_0} ^n \frac{dy}{n_\infty - y} = \int_{t_0} ^t \frac{d x}{\tau}$$
$$n = (n_0 - n_\infty)\exp \Big({\frac{t_0 - t}{\tau}}\Big) + n_\infty$$
Bei dem Gleichgewichtszustand gilt $\frac{dn}{dt} =-(n_0 - n_\infty) \exp ({\frac{t_0 - t}{\tau}}) + n_\infty \overset{!}{=} 0$.
Diese wird bei $t \to \infty$ erreicht, und man erhält:
$$\lim_{t \to \infty} n(t) = n_\infty $$

**Aufgabe (b)**

Die Funktion $n(t)$ ist streng monoton fallend und $n(t) \in [0,1]$, daher muss $n(t_0) = (n_0 - n_\infty) + n_\infty = n_0$ größer oder gleich als der Gleichgewichtszustand $n_\infty$ und kleiner oder gleich als $1$ sein. Daraus folgt für den Anfangswert:
$$n_0 \in [n_\infty , 1]$$
