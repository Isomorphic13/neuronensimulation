```python
import matplotlib.pyplot as plt

from aufgabe2.ode_solver_euler import euler_method
from aufgabe3.receptor_neuron import ReceptorNeuron
from aufgabe3.deciding_neuron import  DecidingNeuron
import numpy as np
from aufgabe3.pixel_sets import get_set_of_pixels

field = np.array([0,1,1,0])
set_of_boards = get_set_of_pixels()


list_of_neurons = []
receptor_neuron0 = ReceptorNeuron(index = 0)
list_of_neurons.append(receptor_neuron0)
receptor_neuron1 = ReceptorNeuron(index = 1)
list_of_neurons.append(receptor_neuron1)
receptor_neuron2 = ReceptorNeuron(index = 2)
list_of_neurons.append(receptor_neuron2)
receptor_neuron3 = ReceptorNeuron(index = 3)
list_of_neurons.append(receptor_neuron3)

weights = [1,-1,-1,1]
deciding_neuron = DecidingNeuron(weights = weights, list_of_neurons = list_of_neurons)

t_a = 0
t_b = 50
dt = 0.01
x_0 = np.array([-65, 0.7, 0.65, 0.6])
tpoints = np.arange(t_a, t_b, dt)

fig, ax = plt.subplots(len(set_of_boards), 1, figsize=(10, 24), sharex=True)
ax = ax.flatten()
fig.suptitle("voltage dynamics in the deciding cell")
i = 0
for i in range(len(set_of_boards)):
    board_label = "".join(str(x) for x in set_of_boards[i])
    ax[i].plot(tpoints, deciding_neuron.get_voltage(set_of_boards[i]), label = board_label)
    ax[i].legend(loc='center left', bbox_to_anchor=(1, 0.5))

plt.show()






```
